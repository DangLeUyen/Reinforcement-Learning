```python
# Download and install Miniconda (force overwrite if exists).
# The '-f' flag handles the 'File or directory already exists' warning.
!wget -q https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh -O miniconda.sh
!chmod +x miniconda.sh
!./miniconda.sh -b -f -p /usr/local

# Initialize conda for the current shell and accept TOS.
# These commands ensure conda is properly set up and its terms are accepted.
!/usr/local/bin/conda init bash
!/usr/local/bin/conda config --set auto_activate_base false
!/usr/local/bin/conda tos accept --override-channels --channel https://repo.anaconda.com/pkgs/main
!/usr/local/bin/conda tos accept --override-channels --channel https://repo.anaconda.com/pkgs/r

# Create a new conda environment named mlagents_env with Python 3.10.12 and ujson.
# Use -y for non-interactive installation.
!/usr/local/bin/conda create -n mlagents_env python=3.10.12 ujson -y

# Set environment variables for Python and conda paths to use mlagents_env.
# This ensures subsequent shell commands and Python imports use the correct environment.
import os
os.environ['PATH'] = "/usr/local/envs/mlagents_env/bin:" + os.environ['PATH']
import sys
sys.path = [p for p in sys.path if not ("site-packages" in p and "/usr/local/envs/mlagents_env" not in p)] # Clean up conflicting paths
sys.path.insert(0, "/usr/local/envs/mlagents_env/lib/python3.10/site-packages")

print("Miniconda and mlagents_env environment setup complete.")
```

### 5. Define the SnowballTarget config file

In ML-Agents, define the training hyperparameters in config.yaml files.

### 6. Train the agent

To train our agent, we need to launch mlagents-learn and select the executable containing the environment.

We define four parameters:

- mlagents-learn <config>: the path where the hyperparameter config file is.
- --env: where the environment executable is.
- --run_id: the name you want to give to your training run id.
- --no-graphics: to not launch the visualization during the training.


```python

# Now, run the mlagents-learn command.
# Explicitly call mlagents-learn from the mlagents_env's bin directory to ensure it's found.
# Added --force flag to overwrite existing run data.
!/usr/local/envs/mlagents_env/bin/mlagents-learn /content/ml-agents/config/ppo/SnowballTarget.yaml --env=./training-envs-executables/linux/SnowballTarget/SnowballTarget --run-id="SnowballTarget1" --no-graphics --force
```
